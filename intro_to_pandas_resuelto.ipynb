## Preparación
Importamos las bibliotecas y reconstruimos el DataFrame original. Las respuestas de limpieza se calculan por separado para conservar los datos originales entre ejercicios.

````python
import pandas as pd
import numpy as np

people_data = {
    'name': ['María', 'Daniel', 'Marisa', 'David', 'Sabrina'],
    'age': [16, 7, 29, None, 33],
    'city': ['Madrid', 'Madrid', 'Málaga', 'Málaga', 'Valladolid'],
    'height': [None, 1.20, 1.60, 2.01, 1.83]
}
people_df = pd.DataFrame(people_data)
print('El DataFrame original contiene 5 personas:')
display(people_df)
````

Toma el dataframe original y guárdalo como un CSV

**Procedimiento:** Se guarda la tabla en un archivo CSV sin añadir el índice como columna.

````python
csv_path = 'data.csv'
people_df.to_csv(csv_path, index=False)
print(f'El DataFrame se ha guardado localmente en {csv_path}.')
````

Ahora lee el archivo CSV

**Procedimiento:** Leemos el CSV generado y mostramos el contenido recuperado.

````python
people_df = pd.read_csv(csv_path)
print('El contenido del archivo CSV es:')
display(people_df)
````

Muestra los 3 primeros registros del DataFrame.

**Procedimiento:** Usamos head(3) para seleccionar las tres primeras filas.

````python
first_three_people = people_df.head(3)
print('Los 3 primeros registros son:')
display(first_three_people)
````

Muestra los 4 últimos registros del DataFrame.

**Procedimiento:** Usamos tail(4) para seleccionar las cuatro últimas filas.

````python
last_four_people = people_df.tail(4)
print('Los 4 últimos registros son:')
display(last_four_people)
````

Dime cuantos datos tiene el DataFrame (💡 es lo mismo que calcular el número de filas). ¿en `numpy` vimos 2 formas de calcular este dato, recuerdas?

**Procedimiento:** Calculamos las filas con len() y con la primera dimensión de shape. No usamos size, que contaría todas las celdas.

````python
row_count = len(people_df)
row_count_from_shape = people_df.shape[0]
print(f'El DataFrame tiene {row_count} filas según len() y {row_count_from_shape} según shape[0].')
````

Muéstrame las estadísticas básicas del DataFrame.

**Procedimiento:** describe() resume las columnas numéricas e ignora los valores nulos al calcular las estadísticas.

````python
summary_statistics = people_df.describe()
print('Las estadísticas básicas de las columnas numéricas son:')
display(summary_statistics)
````

Obtén los datos de la columna 'name' ¿Cual es el tipo de dato de esta columna?

**Procedimiento:** Seleccionamos name. Distinguimos la estructura de la columna (Series) del tipo de sus datos (texto).

````python
names = people_df['name']
print('Los nombres de las personas son:')
display(names)
print(f'La columna es una {type(names).__name__} y su dtype es {names.dtype}; contiene texto.')
````

🌶️ Recuera que una Serie se puede tratar como un array de numpy, por tanto, obtén los datos de la columna "age" y calcula la media. ¿Coincide con el resultado que arroja pandas en la celda superior? (💡 en `numpy` la media se calcula usando `np.mean`)

**Procedimiento:** Convertimos las edades en un array y excluimos los valores nulos antes de usar np.mean(). Así comparamos la media de las edades conocidas con Pandas, que ignora los nulos por defecto.

````python
ages = people_df['age']
valid_ages = ages.dropna().to_numpy()
numpy_mean_age = np.mean(valid_ages)
pandas_mean_age = ages.mean()
print(f'La edad media con NumPy es de {numpy_mean_age:.2f} años.')
print(f'La edad media con Pandas es de {pandas_mean_age:.2f} años.')
print(f'¿Coinciden los resultados? {"Sí" if np.isclose(numpy_mean_age, pandas_mean_age) else "No"}.')
````

Ahora calcula la edad máxima y mínima de la columna "age".

**Procedimiento:** Calculamos el máximo y el mínimo de las edades conocidas.

````python
maximum_age = people_df['age'].max()
minimum_age = people_df['age'].min()
print(f'La edad máxima es de {maximum_age:.0f} años y la mínima es de {minimum_age:.0f} años.')
````

🌶️ Una vez obtenida la edad mínima, ¿Cual es el nombre de dicha persona?

**Procedimiento:** idxmin() devuelve el índice de la edad mínima. Usamos loc para consultar el nombre de esa fila.

````python
youngest_person_index = people_df['age'].idxmin()
youngest_person_name = people_df.loc[youngest_person_index, 'name']
print(f'La persona de menor edad se llama {youngest_person_name}.')
````

Filtra el DataFrame: muestra las filas donde la edad sea mayor a 25.

**Procedimiento:** Aplicamos una condición para seleccionar edades superiores a 25.

````python
older_people = people_df[people_df['age'] > 25]
print('Las personas mayores de 25 años son:')
display(older_people)
````

Filtra el DataFrame: muestra las filas donde la gente sea de 'Málaga'. (💡 utiliza el operador `==`)

**Procedimiento:** Comparamos la ciudad con Málaga mediante ==.

````python
malaga_people = people_df[people_df['city'] == 'Málaga']
print('Las personas de Málaga son:')
display(malaga_people)
````

🌶️ Filtra el Dataframe: filas donde la gente sea de 'Málaga' y mayor de 25. (💡 para sumar dos condiciones se utiliza el operador `and` que se representa en python con `&`, i.e.: 

```python
compose_condition = condition1 & condition2
```

**Procedimiento:** Combinamos las dos condiciones con & y ponemos cada condición entre paréntesis. La edad desconocida de David no cumple el filtro.

````python
malaga_and_over_25 = (people_df['city'] == 'Málaga') & (people_df['age'] > 25)
older_malaga_people = people_df[malaga_and_over_25]
print('Las personas de Málaga mayores de 25 años son:')
display(older_malaga_people)
````

Obtén la máscara de los valores que son nulos en el DataFrame.

**Procedimiento:** isnull() devuelve True donde hay un dato ausente.

````python
missing_values_mask = people_df.isnull()
print('La máscara indica con True los valores nulos:')
display(missing_values_mask)
````

Ahora obtén la máscara de los valores que NO son nulos en el DataFrame. (💡 busca en la documentación de pandas, o en google, cómo hacerlo)

**Procedimiento:** notnull() devuelve True donde hay un dato presente. También se puede usar notna().

````python
non_missing_values_mask = people_df.notnull()
print('La máscara indica con True los valores que no son nulos:')
display(non_missing_values_mask)
````

Elimina los valores nulos del DataFrame.

**Procedimiento:** dropna() elimina las filas que contienen algún nulo. Guardamos el resultado aparte para mantener el DataFrame original.

````python
complete_people_df = people_df.dropna()
print(f'Tras eliminar las filas con valores nulos, quedan {len(complete_people_df)} personas:')
display(complete_people_df)
````

Rellena todos los valores nulos con 1.

**Procedimiento:** Rellenamos los valores ausentes con 1 y guardamos el resultado en otra variable.

````python
filled_people_df = people_df.fillna(1)
print('El DataFrame con todos los valores nulos sustituidos por 1 es:')
display(filled_people_df)
````

Ahora rellena sólo los valores nulos de la columna 'age' con 10.

**Procedimiento:** fillna() con un diccionario permite rellenar únicamente la columna age.

````python
age_filled_people_df = people_df.fillna({'age': 10})
print('El DataFrame con la edad desconocida sustituida por 10 es:')
display(age_filled_people_df)
````

🌶️ Ahora rellena los valores nulos de la columna 'height' con la media de dicha columna

**Procedimiento:** Calculamos la media de las alturas conocidas y rellenamos solo los valores ausentes de height.

````python
mean_height = people_df['height'].mean()
height_filled_people_df = people_df.fillna({'height': mean_height})
print(f'La altura media es {mean_height:.2f} m; sustituimos por ella la altura desconocida:')
display(height_filled_people_df)
````

## BONUS: Explora los datos del dataset de Titanic

El dataset se lee directamente desde su URL. Contiene información sobre los pasajeros, como edad, clase y supervivencia.

````python
titanic_url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic_df = pd.read_csv(titanic_url)
print(f'El dataset contiene {len(titanic_df)} pasajeros. Sus primeros 5 registros son:')
display(titanic_df.head(5))
````

**¿Qué porcentaje de pasajeros sobrevivió?**

**Procedimiento:** Survived vale 1 para supervivientes y 0 para el resto. Su media multiplicada por 100 proporciona el porcentaje.

````python
survival_rate = titanic_df['Survived'].mean() * 100
print(f'La tasa de supervivencia es del {survival_rate:.2f}%.')
````

**¿Cuál es la edad media de los pasajeros?**

**Procedimiento:** Calculamos la media de las edades conocidas; los valores ausentes se omiten.

````python
mean_passenger_age = titanic_df['Age'].mean()
print(f'La edad media de los pasajeros con edad conocida es de {mean_passenger_age:.2f} años.')
````

**¿Cuántos pasajeros había en cada clase?**

**Procedimiento:** Contamos los valores de Pclass y ordenamos por clase.

````python
passengers_by_class = titanic_df['Pclass'].value_counts().sort_index()
for passenger_class, passenger_count in passengers_by_class.items():
    print(f'En la clase {passenger_class} había {passenger_count} pasajeros.')
````

**¿Cuál es la edad media de los pasajeros que sobrevivieron y de los que no?**

**Procedimiento:** Agrupamos por supervivencia y calculamos la media de las edades conocidas de cada grupo.

````python
mean_age_by_survival = titanic_df.groupby('Survived')['Age'].mean()
print(f'La edad media de quienes sobrevivieron es de {mean_age_by_survival.loc[1]:.2f} años.')
print(f'La edad media de quienes no sobrevivieron es de {mean_age_by_survival.loc[0]:.2f} años.')
````

**¿Cuál es la edad máxima de los pasajeros que sobrevivieron y de los que no?**

**Procedimiento:** Agrupamos por supervivencia y obtenemos la mayor edad conocida de cada grupo.

````python
maximum_age_by_survival = titanic_df.groupby('Survived')['Age'].max()
print(f'La edad máxima de quienes sobrevivieron es de {maximum_age_by_survival.loc[1]:.0f} años.')
print(f'La edad máxima de quienes no sobrevivieron es de {maximum_age_by_survival.loc[0]:.0f} años.')
````
